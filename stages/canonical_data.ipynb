### Catalog bootstrap

Expects an existing UC catalog (create it in the workspace first).
Creates the `metadata` schema this demo writes into.
Kitchen/menu/history tables are created by later stages.


```python
%pip install --upgrade databricks-sdk
```

```python
dbutils.library.restartPython()
```

```python
CATALOG = dbutils.widgets.get("CATALOG")
try:
    SIMULATOR_SCHEMA = dbutils.widgets.get("SIMULATOR_SCHEMA") or "metadata"
except Exception:
    SIMULATOR_SCHEMA = "metadata"

try:
    spark.sql(f"DESCRIBE CATALOG `{CATALOG}`")
except Exception as e:
    raise RuntimeError(
        f"Catalog `{CATALOG}` does not exist or is not visible from this workspace. "
        "Create it first (Catalog Explorer → Create catalog), then re-run."
    ) from e
try:
    spark.sql(f"GRANT USE CATALOG ON CATALOG `{CATALOG}` TO `account users`")
except Exception as e:
    print(f"GRANT USE CATALOG skipped: {e}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS `{CATALOG}`.`{SIMULATOR_SCHEMA}`")
print(f"Catalog {CATALOG}, schema {SIMULATOR_SCHEMA}")
```
